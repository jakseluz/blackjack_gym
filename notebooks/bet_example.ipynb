# Blackjack4game-v1 -> version with betting

### Random game example

```python
import gymnasium as gym
import blackjack_env
import time

env = gym.make("Blackjack4game-v1", render_mode="terminal")
done = False
obs, info = env.reset()

# betting
# bet: 0.0 - 1.0 (percentage)
# game_decision: 1.0 - 4.99 (almost like in v0, but float to fit the networks)
action = env.action_space.sample()  # [bet, game_decision]
obs, reward, terminated, truncated, info = env.step(action)

# playing
while not done:
    action = env.action_space.sample()  # [bet, game_decision]
    game_decision = int(action[1])
    while info["action_mask"][game_decision] == 0:
        action = env.action_space.sample()
        game_decision = int(action[1])
    obs, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    time.sleep(3)
    if done:
        print(info)
        print(f"Reward: {reward}")

time.sleep(7)
env.close()
```
